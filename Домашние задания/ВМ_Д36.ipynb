Необходимо составить максимально лаконичный код, который строил бы по заданной выборке доверительный интервал для математического ожидания и позволял бы пользователю выбрать тип доверительного интервала:
1) асимптотический доверительный интервал на квантилях стандартного нормального закона распределения,
2) доверительный интервал на коэффициентах Стьюдента,
3) доверительный интервал Хора-Хора для ограниченных распределений.

```python
import numpy as np
from scipy.stats import norm, t


def interval(x, q=0.95, method='normal', bounds=None):
    # Возможные method - normal, student, hora
    x = np.asarray(x, dtype=float)
    n = len(x)
    mean, s = x.mean(), x.std(ddof=1)

    if method == 'normal':
        d = (s / np.sqrt(n)) * norm.ppf((1 + q) / 2)
    elif method == 'student':
        d = (s / np.sqrt(n)) * t.ppf((1 + q) / 2, n - 1)
    elif method == 'hora':
        if bounds is None:
            a, b = min(x), max(x)
        else:
            a, b = bounds
        d = (b - a) * (np.sqrt(-np.log((1 + q) / 2) / (2 * n)) - (1 / (6 * n)))

    return mean - d, mean + d


x = [1.2, 2.3, 2.7, 2.1, 2.6, 3.1, 1.8, 3.0, 1.7, 1.4]
print(*interval(x, 0.95, 'normal'))
print(*interval(x, 0.95, 'student'))
print(*interval(x, 0.95, 'hora', bounds=(1, 4)))
```
