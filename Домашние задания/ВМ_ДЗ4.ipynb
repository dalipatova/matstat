Необходимо составить максимально лаконичный код, который рассчитывал бы по заданной выборке следующие точечные оценки:
1) математического ожидания и медианы,
2) дисперсии и среднеквадратического отклонения (два варианта - смещенную и несмещенную оценку при использовании в качестве оценки математического ожидания выборочного среднего),
3) несмещенную оценку коэффициентов асимметрии и эксцесса.

```python
import pandas as pd

sample = pd.Series([1, 1, 4, 3, 2, 8, 3, 9, 0, 4, 6, 5])
print('Математическое ожидание:', float(sample.mean()))
print('Медиана:', float(sample.median()))
print('Смещенная дисперсия:', float(sample.var(ddof=0)))
print('Несмещенная дисперсия:', float(sample.var(ddof=1)))
print('Смещенное среднеквадратическое отклонение:', float(sample.std(ddof=0)))
print('Несмещенное среднеквадратическое отклонение:', float(sample.std(ddof=1)))
print('Коэффициент асимметрии:', float(sample.skew()))
print('Коэффициент эксцесса:', float(sample.kurtosis()))
```
