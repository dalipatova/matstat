Необходимо составить максимально лаконичный код, который строил бы по заданной выборке методом бутстреп доверительный интервал для оценки, вычисляемой по отдельной заданной процедуре (смысл задания аналогичен заданию на метод максимального правдоподобия).

Спецификация программы предполагается следующей: есть процедура, реализующая построение границ доверительного интервала для заданной выборки методом бутстреп, условно назовем ее cint = bootci(@est, x).

Ей на вход передается указатель или ссылка @est на функцию est, которая реализует вычисление оценки по заданной выборке x: p_est = est(x), где x - вектор элементов выборки, p_est - возвращаемое значение оценки интересующего параметра.

Переменная cint представляет собой вектор из двух элементов, являющихся границами доверительного интервала.

```python
import numpy as np

# B - количество генерируемых выборок
def bootci(est, x, B=1000):
    p = np.array([est(np.random.choice(x, len(x), True)) for _ in range(B)])
    return np.percentile(p, [2.5, 97.5]) # 95%-й доверительный интервал

# Например, оценка среднего
x = np.array([1.2, 0.8, 1.5, 1.1, 0.9])
print(*bootci(np.mean, x))
```
